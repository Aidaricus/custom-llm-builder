```python
%load_ext autoreload
%autoreload 2

import sys
import asyncio
from pathlib import Path

project_root = Path.cwd().parent
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

from src.llm_platform.data_foundry.evaluator_data import DatasetEvaluator

# 1. Инициализируем оценщика
# Используем ту же модель для тестов, можно поменять на что-то более мощное (например, GPT-4o), если позволяет API
evaluator = DatasetEvaluator(
    model_name="nvidia/nemotron-3-nano-30b-a3b:free",
    max_concurrent_requests=2
)

# 2. Настраиваем пути
sanitized_sft = project_root / "data/sanitized/sanitized_sft_dataset.jsonl"
chunks_file = project_root / "chunks.jsonl" # Убедись, что этот файл существует!
report_file = project_root / "dataset_health.md"

# 3. Передаем статистику, которую мы получили на этапе очистки
fake_sanitization_stats = {
    "total_processed": 81,
    "passed": 80,
    "rejected_structure": 1,
    "rejected_length": 0,
    "rejected_refusal": 0,
    "rejected_duplicate": 0
}

print("Запускаем LLM-судью. Это может занять пару минут, так как мы ждем ответы от API...")

# 4. Запускаем оценку
await evaluator.run_evaluation(
    sanitized_sft_file=sanitized_sft,
    chunks_file=chunks_file,
    report_output_file=report_file,
    sanitization_stats=fake_sanitization_stats
)

print(f"Готово! Отчет сохранен в: {report_file}")
```
