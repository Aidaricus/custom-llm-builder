```python
%load_ext autoreload
%autoreload 2

import sys
from pathlib import Path
import json

project_root = Path.cwd().parent
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

from src.llm_platform.data_foundry.dpo_pipeline import DPOPipeline
```

```python
# На вход подаем наш усложненный датасет
input_dataset = project_root / "data/processed/test_sft_dataset.jsonl"
output_dataset = project_root / "data/processed/test_dpo_dataset.jsonl"
prompts_file = project_root / "src/llm_platform/data_foundry/dpo_prompts.yaml"

pipeline = DPOPipeline(
    input_file=input_dataset,
    output_file=output_dataset,
    prompts_file=prompts_file,
    model_name="nvidia/nemotron-3-nano-30b-a3b:free", # или любая другая стабильная
    max_concurrent_requests=3,
)

await pipeline.run_generation()
```

```python
with open(output_dataset, "r", encoding="utf-8") as f:
    first_line = f.readline()
    if first_line:
        triplet = json.loads(first_line)
        
        print("=== PROMPT ===")
        print(triplet['prompt'][:500] + "...\n")
        
        print("=== CHOSEN (Идеальный ответ) ===")
        print(triplet['chosen'][:500] + "...\n")
        
        print(f"=== REJECTED (Плохой ответ) | Причина: {triplet['reject_reason']} ===")
        print(triplet['rejected'][:500] + "...\n")
    else:
        print("Файл пуст.")
```
